```python
import polars as pl
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
data =[
    'data1_avg.csv','data2_avg.csv','data3_avg.csv','data4_avg.csv','data5_avg.csv','data6_avg.csv','data7_avg.csv','data8_avg.csv',
    'data9_avg.csv','data10_avg.csv','data11_avg.csv','data12_avg.csv','data13_avg.csv','data14_avg.csv','data15_avg.csv','data16_avg.csv',
    'data17_avg.csv','data18_avg.csv','data19_avg.csv','data20_avg.csv','data21_avg.csv','data22_avg.csv','data23_avg.csv','data24_avg.csv',
    'data25_avg.csv','data26_avg.csv','data27_avg.csv'
]
plt.figure(figsize=(20, 25))
for i in range(len(data)):
    plt.subplot(7, 4, i+1)
    df = pd.read_csv(data[i])
    plt.plot(df.Number,df['P1[Avg(?)]'])
    plt.scatter(df.Number,df['P1[Avg(?)]'])
    plt.title('Teamperaturs Vs Time for pass')
    plt.xlabel("Time")
    plt.ylabel('Temperature')
plt.tight_layout()
plt.show()
```
